# Escalado
Conjunto de datos: iris (incluido en sklearn)
Este conjunto de datos es compacto y perfecto para demostrar técnicas de escalado.

```python
# Importar librerías necesarias
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler, StandardScaler
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Cargar el conjunto de datos
iris = load_iris(as_frame=True)
data = iris.data
```

```python
# Visualizar las estadísticas iniciales
print("Estadísticas iniciales:")
data.describe()
```

```python
# Aplicar MinMaxScaler
minmax_scaler = MinMaxScaler()
data_minmax = pd.DataFrame(minmax_scaler.fit_transform(data), columns=data.columns)
```

```python
# Aplicar StandardScaler
standard_scaler = StandardScaler()
data_standard = pd.DataFrame(standard_scaler.fit_transform(data), columns=data.columns)
```

### 1. ¿Qué diferencias encuentras entre las estadísticas originales y las del escalado MinMax?


```python
original_stats = data.describe()
minmax_stats = data_minmax.describe()

diferencias = original_stats, minmax_stats
final_stats = pd.concat([original_stats, minmax_stats], axis=1, keys=['Original', 'MinMaxScaler'])

print("\nDiferencias entre estadísticas iniciales y MinMaxScaler:")
print(diferencias)
print("\n Estadisticas finales: ")
print(final_stats)


"""Las diferencias a simple vista son: el rango en que se encuentran los datos, 
ya que con minmax los datos se engloban entre 0 y 1.
"""
```

### 2. Crea una gráfica de la distribución de la variable 'sepal length (cm)' antes y después del escalado.

```python
# grafica de distribución de los datos originales y escalados  

for sepal_length in data.columns:
    plt.figure(figsize=(12, 4))
    
    plt.subplot(1, 3, 1)
    plt.hist(data[sepal_length], bins=20, color='blue', alpha=0.7)
    plt.title(f'Original {sepal_length}')
    
    plt.subplot(1, 3, 2)
    plt.hist(data_minmax[sepal_length], bins=20, color='green', alpha=0.7)
    plt.title(f'MinMax Scaled {sepal_length}')

    plt.tight_layout()
    plt.show()
```

### 3. Crea una gráfica de dispersión entre 'sepal length (cm)' y 'sepal width (cm)' con los datos escalados por StandardScaler.

```python
#grafica de dispersión entre sepal_length y sepal_width para los datos originales y escalados
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.scatter(data['sepal length (cm)'], data['sepal width (cm)'], color='blue', alpha=0.7)
plt.title('Original Data')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')

plt.subplot(1, 2, 2)
plt.scatter(data_standard['sepal length (cm)'], data_standard['sepal width (cm)'], color='red', alpha=0.7)
plt.title('Standard Scaled Data')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)') 


plt.show()
```
