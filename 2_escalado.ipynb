# Escalado
Conjunto de datos: iris (incluido en sklearn)
Este conjunto de datos es compacto y perfecto para demostrar técnicas de escalado.

```python
# Importar librerías necesarias
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler, StandardScaler
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Cargar el conjunto de datos
iris = load_iris(as_frame=True)
data = iris.data
```

```python
# Aplicar MinMaxScaler
minmax_scaler = MinMaxScaler()
data_minmax = pd.DataFrame(minmax_scaler.fit_transform(data), columns=data.columns)
```

```python
data_minmax.describe()
```

```python
# Aplicar StandardScaler
standard_scaler = StandardScaler()
data_standard = pd.DataFrame(standard_scaler.fit_transform(data), columns=data.columns)
```

```python
# ==============================================================================
# 1. Comparación de Estadísticas Descriptivas (Original vs MinMaxScaler)
# ==============================================================================
print("=== 1. Estadísticas de los datos originales ===")
display(data.describe())

print("\n=== 2. Estadísticas con MinMaxScaler ===")
display(data_minmax.describe())

# Tabla comparativa resumida (Mínimo, Máximo, Media y Desviación Estándar)
resumen_comparativo = pd.DataFrame({
    'Min (Original)': data.min(),
    'Min (MinMax)': data_minmax.min(),
    'Max (Original)': data.max(),
    'Max (MinMax)': data_minmax.max(),
    'Media (Original)': data.mean().round(3),
    'Media (MinMax)': data_minmax.mean().round(3),
    'Std (Original)': data.std().round(3),
    'Std (MinMax)': data_minmax.std().round(3)
})
print("\n=== Resumen Comparativo de Métricas Clave ===")
display(resumen_comparativo)

# ==============================================================================
# Gráfica 1: Distribución de los datos antes del escalado (Original)
# ==============================================================================
fig1, axes1 = plt.subplots(1, 2, figsize=(14, 4.5))

# Densidad de probabilidad (KDE)
data.plot(kind='kde', ax=axes1[0], linewidth=2)
axes1[0].set_title('Densidad de Probabilidad (KDE)')
axes1[0].set_xlabel('Valor en cm (Escala original)')
axes1[0].set_ylabel('Densidad')
axes1[0].grid(True, linestyle='--', alpha=0.5)

# Diagrama de caja (Boxplot)
data.boxplot(ax=axes1[1], patch_artist=True)
axes1[1].set_title('Diagrama de Caja (Boxplot)')
axes1[1].set_ylabel('Valor en cm (Escala original)')
axes1[1].tick_params(axis='x', rotation=15)
axes1[1].grid(True, linestyle='--', alpha=0.5)

plt.suptitle('1. Distribución de los Datos Originales (Antes del escalado)', fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

# ==============================================================================
# Gráfica 2: Distribución de los datos con MinMaxScaler
# ==============================================================================
fig2, axes2 = plt.subplots(1, 2, figsize=(14, 4.5))

# Densidad de probabilidad (KDE)
data_minmax.plot(kind='kde', ax=axes2[0], linewidth=2)
axes2[0].set_title('Densidad de Probabilidad (KDE)')
axes2[0].set_xlabel('Valor escalado en el rango [0, 1]')
axes2[0].set_ylabel('Densidad')
axes2[0].grid(True, linestyle='--', alpha=0.5)

# Diagrama de caja (Boxplot)
data_minmax.boxplot(ax=axes2[1], patch_artist=True)
axes2[1].set_title('Diagrama de Caja (Boxplot)')
axes2[1].set_ylabel('Valor escalado [0, 1]')
axes2[1].tick_params(axis='x', rotation=15)
axes2[1].grid(True, linestyle='--', alpha=0.5)

plt.suptitle('2. Distribución de los Datos con MinMaxScaler (Rango [0, 1])', fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

# ==============================================================================
# Gráfica 3: Distribución de los datos con StandardScaler
# ==============================================================================
fig3, axes3 = plt.subplots(1, 2, figsize=(14, 4.5))

# Densidad de probabilidad (KDE)
data_standard.plot(kind='kde', ax=axes3[0], linewidth=2)
axes3[0].set_title('Densidad de Probabilidad (KDE)')
axes3[0].set_xlabel('Valor estandarizado (z-score)')
axes3[0].set_ylabel('Densidad')
axes3[0].grid(True, linestyle='--', alpha=0.5)

# Diagrama de caja (Boxplot)
data_standard.boxplot(ax=axes3[1], patch_artist=True)
axes3[1].set_title('Diagrama de Caja (Boxplot)')
axes3[1].set_ylabel('Valor estandarizado (z-score)')
axes3[1].tick_params(axis='x', rotation=15)
axes3[1].grid(True, linestyle='--', alpha=0.5)

plt.suptitle('3. Distribución de los Datos con StandardScaler (Media = 0, Desv. Est. = 1)', fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

#### Conclusión y Diferencias observadas entre los datos Originales y MinMaxScaler:

1. **Rango y límites de valores (`min` y `max`):**
   - **Datos originales:** Cada característica conserva sus unidades naturales (cm) y cuenta con rangos muy distintos entre sí. Por ejemplo, `petal width (cm)` tiene un mínimo de 0.1 y un máximo de 2.5 (rango de 2.4 cm), mientras que `sepal length (cm)` oscila entre 4.3 y 7.9 (rango de 3.6 cm).
   - **MinMaxScaler:** Todas las variables son comprimidas de forma estricta al intervalo cerrado **[0, 1]**. Para **todas** las variables sin excepción, el valor mínimo es exactamente **0.0** y el valor máximo es exactamente **1.0**.

2. **Tendencia central (`mean`):**
   - **Datos originales:** Las medias varían ampliamente según la magnitud de cada variable (desde 1.20 en `petal width` hasta 5.84 en `sepal length`).
   - **MinMaxScaler:** Las medias se recalculan de manera proporcional dentro de [0, 1]. Ahora todas quedan en un rango homogéneo (entre aprox 0.43 y 0.47).

3. **Dispersión (`std`):**
   - **Datos originales:** Las desviaciones estándar son heterogéneas (0.44 a 1.77), reflejando diferencias absolutas de variabilidad entre características.
   - **MinMaxScaler:** Las desviaciones estándar disminuyen drásticamente (entre aprox 0.18 y 0.32), reflejando la contracción de la escala, pero conservando la dispersión relativa de los datos.

4. **Preservación de la forma de la distribución:**
   - La transformación MinMax es estrictamente lineal ($x' = \frac{x - x_{min}}{x_{max} - x_{min}}$). Por lo tanto, **la forma geométrica de la distribución, la asimetría (skewness) y las distancias relativas entre los puntos se conservan exactamente iguales**; únicamente cambia la escala numérica en los ejes.

5. **Comparación con StandardScaler:**
   - Mientras que MinMaxScaler acota los datos rígidamente entre 0 y 1, **StandardScaler** no fija un rango acotado, sino que desplaza los datos para tener media 0 ($\mu \approx 0$) y reescala la dispersión para que la desviación estándar sea 1 ($\sigma = 1$).


### 2. Crea una gráfica de la distribución de la variable 'sepal length (cm)' antes y después del escalado.

```python
# ==============================================================================
# Gráfica de distribución de 'sepal length (cm)': Antes y Después del Escalado
# ==============================================================================
import numpy as np
from scipy.stats import norm

variable = 'sepal length (cm)'

# Parámetros teóricos de la distribución normal para los datos originales
mu_orig = data[variable].mean()
std_orig = data[variable].std()
x_orig = np.linspace(data[variable].min() - 0.5, data[variable].max() + 0.5, 200)

# ------------------------------------------------------------------------------
# 1. Antes vs Después con MinMaxScaler
# ------------------------------------------------------------------------------
mu_mm = data_minmax[variable].mean()
std_mm = data_minmax[variable].std()
x_mm = np.linspace(-0.1, 1.1, 200)

fig1, axes1 = plt.subplots(1, 2, figsize=(14, 4.5))

# Antes (Original)
axes1[0].hist(data[variable], bins=15, density=True, alpha=0.6, color='steelblue', edgecolor='black', label='Histograma')
data[variable].plot(kind='kde', ax=axes1[0], color='darkblue', linewidth=2, label='Densidad (KDE)')
axes1[0].plot(x_orig, norm.pdf(x_orig, mu_orig, std_orig), 'r--', linewidth=2, label=f'Normal N({mu_orig:.2f}, {std_orig:.2f})')
axes1[0].set_title(f'Antes del Escalado (Original)\nMedia = {mu_orig:.2f}, Desv = {std_orig:.2f}')
axes1[0].set_xlabel('sepal length (cm)')
axes1[0].set_ylabel('Densidad')
axes1[0].legend()
axes1[0].grid(True, linestyle='--', alpha=0.5)

# Después (MinMaxScaler)
axes1[1].hist(data_minmax[variable], bins=15, density=True, alpha=0.6, color='darkorange', edgecolor='black', label='Histograma')
data_minmax[variable].plot(kind='kde', ax=axes1[1], color='brown', linewidth=2, label='Densidad (KDE)')
axes1[1].plot(x_mm, norm.pdf(x_mm, mu_mm, std_mm), 'r--', linewidth=2, label=f'Normal N({mu_mm:.2f}, {std_mm:.2f})')
axes1[1].set_title(f'Después del Escalado (MinMaxScaler)\nMedia = {mu_mm:.2f}, Desv = {std_mm:.2f}')
axes1[1].set_xlabel('sepal length escalado [0, 1]')
axes1[1].set_ylabel('Densidad')
axes1[1].legend()
axes1[1].grid(True, linestyle='--', alpha=0.5)

plt.suptitle("Distribución de 'sepal length (cm)': Antes vs Después (MinMaxScaler)", fontsize=13, fontweight='bold', y=1.03)
plt.tight_layout()
plt.show()

# ------------------------------------------------------------------------------
# 2. Antes vs Después con StandardScaler
# ------------------------------------------------------------------------------
mu_std = data_standard[variable].mean()
std_scale = data_standard[variable].std()
x_std = np.linspace(-3.5, 3.5, 200)

fig2, axes2 = plt.subplots(1, 2, figsize=(14, 4.5))

# Antes (Original)
axes2[0].hist(data[variable], bins=15, density=True, alpha=0.6, color='steelblue', edgecolor='black', label='Histograma')
data[variable].plot(kind='kde', ax=axes2[0], color='darkblue', linewidth=2, label='Densidad (KDE)')
axes2[0].plot(x_orig, norm.pdf(x_orig, mu_orig, std_orig), 'r--', linewidth=2, label=f'Normal N({mu_orig:.2f}, {std_orig:.2f})')
axes2[0].set_title(f'Antes del Escalado (Original)\nMedia = {mu_orig:.2f}, Desv = {std_orig:.2f}')
axes2[0].set_xlabel('sepal length (cm)')
axes2[0].set_ylabel('Densidad')
axes2[0].legend()
axes2[0].grid(True, linestyle='--', alpha=0.5)

# Después (StandardScaler)
axes2[1].hist(data_standard[variable], bins=15, density=True, alpha=0.6, color='seagreen', edgecolor='black', label='Histograma')
data_standard[variable].plot(kind='kde', ax=axes2[1], color='darkgreen', linewidth=2, label='Densidad (KDE)')
axes2[1].plot(x_std, norm.pdf(x_std, 0, 1), 'r--', linewidth=2, label='Normal Estándar N(0, 1)')
axes2[1].set_title(f'Después del Escalado (StandardScaler)\nMedia ≈ {mu_std:.2f}, Desv = {std_scale:.2f}')
axes2[1].set_xlabel('sepal length estandarizado (z-score)')
axes2[1].set_ylabel('Densidad')
axes2[1].legend()
axes2[1].grid(True, linestyle='--', alpha=0.5)

plt.suptitle("Distribución de 'sepal length (cm)': Antes vs Después (StandardScaler)", fontsize=13, fontweight='bold', y=1.03)
plt.tight_layout()
plt.show()
```

#### Observaciones sobre la distribución de 'sepal length (cm)':

1. **Conservación de la forma acampanada:**
   - En ambas transformaciones (MinMaxScaler y StandardScaler), **la forma de la distribución se mantiene idéntica a la original**. Ambas son transformaciones lineales que no modifican la asimetría ni la curvatura de los datos.

2. **Efecto de MinMaxScaler:**
   - Traslada y contrae el rango original de $[4.3, 7.9]$ cm al intervalo acotado **$[0, 1]$**.
   - La media pasa de $5.84$ a $0.43$, y la desviación estándar disminuye de $0.83$ a $0.23$.

3. **Efecto de StandardScaler:**
   - Convierte la variable en una **distribución Normal Estándar $\mathcal{N}(0, 1)$**.
   - Centra la media exactamente en $0$ ($\mu \approx 0$) y ajusta la escala para que la desviación estándar sea exactamente $1$ ($\sigma = 1$).
   - La curva normal teórica $\mathcal{N}(0, 1)$ se superpone perfectamente sobre el histograma y la curva KDE.

### 4. ¿Qué técnica de escalado crees que es más adecuada para este conjunto de datos? ¿Por qué?

```python
# ==============================================================================
# 4. Comparativa técnica y experimental: MinMaxScaler vs StandardScaler en Iris
# ==============================================================================
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC

# 1. Diagnóstico estadístico de outliers (regla 1.5 * IQR) y asimetría
diagnostico = []
for col in data.columns:
    q1 = data[col].quantile(0.25)
    q3 = data[col].quantile(0.75)
    iqr = q3 - q1
    outliers = ((data[col] < (q1 - 1.5 * iqr)) | (data[col] > (q3 + 1.5 * iqr))).sum()
    diagnostico.append({
        'Variable': col,
        'Outliers (IQR)': int(outliers),
        'Asimetría (Skew)': round(data[col].skew(), 3),
        'Std (MinMax)': round(data_minmax[col].std(), 3),
        'Std (Standard)': round(data_standard[col].std(), 3)
    })

df_diag = pd.DataFrame(diagnostico)
print("=== Diagnóstico de distribución y varianza en Iris ===")
display(df_diag)

# 2. Evaluación experimental en modelos sensibles a la escala (sin fuga de datos vía Pipeline)
X_iris = iris.data
y_iris = iris.target
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

modelos = {
    'KNN (k=5)': KNeighborsClassifier(n_neighbors=5),
    'SVM (RBF)': SVC(kernel='rbf', random_state=42),
    'Regresión Logística': LogisticRegression(max_iter=1000, random_state=42)
}

escaladores = {
    'Sin escalar': None,
    'MinMaxScaler': MinMaxScaler(),
    'StandardScaler': StandardScaler()
}

resultados = []
for nombre_m, modelo in modelos.items():
    for nombre_e, escalador in escaladores.items():
        pasos = [('clf', modelo)] if escalador is None else [('scaler', escalador), ('clf', modelo)]
        pipe = Pipeline(pasos)
        scores = cross_val_score(pipe, X_iris, y_iris, cv=cv, scoring='accuracy')
        resultados.append({
            'Modelo': nombre_m,
            'Escalado': nombre_e,
            'Accuracy': scores.mean()
        })

df_res = pd.DataFrame(resultados)
tabla_res = df_res.pivot(index='Modelo', columns='Escalado', values='Accuracy')[['Sin escalar', 'MinMaxScaler', 'StandardScaler']]
print("\n=== Accuracy promedio (CV 5-folds) según técnica de escalado ===")
display(tabla_res.round(4))

# 3. Gráfica comparativa de varianza post-escalado y desempeño de modelos
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

x_pos = range(len(df_diag))
width = 0.35
axes[0].bar([p - width/2 for p in x_pos], df_diag['Std (MinMax)'], width=width, color='darkorange', alpha=0.85, label='MinMaxScaler')
axes[0].bar([p + width/2 for p in x_pos], df_diag['Std (Standard)'], width=width, color='seagreen', alpha=0.85, label='StandardScaler')
axes[0].set_xticks(list(x_pos))
axes[0].set_xticklabels(df_diag['Variable'], rotation=15)
axes[0].set_ylabel('Desviación Estándar')
axes[0].set_title('Dispersión resultante por variable')
axes[0].legend()
axes[0].grid(axis='y', linestyle='--', alpha=0.5)

tabla_res.plot(kind='bar', ax=axes[1], color=['gray', 'darkorange', 'seagreen'], alpha=0.85, rot=10)
axes[1].set_ylim(min(0.85, df_res['Accuracy'].min() - 0.03), 1.01)
axes[1].set_ylabel('Accuracy promedio (CV)')
axes[1].set_title('Desempeño en modelos sensibles a la escala')
axes[1].legend(title='Técnica')
axes[1].grid(axis='y', linestyle='--', alpha=0.5)

plt.suptitle('Comparación técnica y empírica: MinMaxScaler vs StandardScaler en Iris', fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

#### Conclusión sobre qué técnica es más adecuada para Iris:

1. **Técnica recomendada en general: `StandardScaler`**
   - **Varianza uniforme ($\sigma = 1$):** Como muestra la primera gráfica, `StandardScaler` garantiza que las 4 características tengan exactamente la misma desviación estándar ($\sigma = 1$) y media $0$, mientras que `MinMaxScaler` conserva desviaciones estándar distintas entre sépalos y pétalos.
   - **Menor sensibilidad a valores extremos:** `MinMaxScaler` depende únicamente del mínimo y máximo observados; ante valores atípicos (como los detectados por IQR en `'sepal width (cm)'`), comprime al resto de los puntos en una franja estrecha. `StandardScaler` usa $\mu$ y $\sigma$ de toda la muestra.
   - **Naturaleza biológica de los datos:** Las dimensiones de las flores no tienen cotas físicas obligatorias en $[0, 1]$ y se adaptan bien a una estandarización *z-score*.

2. **¿Cuándo preferir `MinMaxScaler` o no escalar?**
   - **`MinMaxScaler`** conviene si el algoritmo exige valores estrictamente acotados en $[0, 1]$ (por ejemplo, ciertas **Redes Neuronales**).
   - **Sin escalado:** En modelos basados en árboles (**Decision Tree, Random Forest**), los cortes por umbral (`x <= c`) no se ven afectados por transformaciones lineales crecientes.
