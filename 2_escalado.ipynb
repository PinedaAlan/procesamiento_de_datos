# Escalado
Conjunto de datos: iris (incluido en sklearn)
Este conjunto de datos es compacto y perfecto para demostrar técnicas de escalado.

```python
# Importar librerías necesarias
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler, StandardScaler
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Cargar el conjunto de datos
iris = load_iris(as_frame=True)
data = iris.data
```

```python
# Visualizar las estadísticas iniciales
print("Estadísticas iniciales:")
data.describe()
```

```python
# Aplicar MinMaxScaler
minmax_scaler = MinMaxScaler()
data_minmax = pd.DataFrame(minmax_scaler.fit_transform(data), columns=data.columns)
```

```python
# Aplicar StandardScaler
standard_scaler = StandardScaler()
data_standard = pd.DataFrame(standard_scaler.fit_transform(data), columns=data.columns)
```

### 2. Crea una gráfica de la distribución de la variable 'sepal length (cm)' antes y después del escalado.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.histplot(data['sepal length (cm)'], kde=True, ax=axes[0], color='skyblue')
axes[0].set_title('Original: Sepal Length')

sns.histplot(data_minmax['sepal length (cm)'], kde=True, ax=axes[1], color='salmon')
axes[1].set_title('MinMax Scaled: Sepal Length')

plt.tight_layout()
plt.show()
```

### 3. Crea una gráfica de dispersión entre 'sepal length (cm)' y 'sepal width (cm)' con los datos escalados por StandardScaler.

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(
    x=data_standard['sepal length (cm)'], 
    y=data_standard['sepal width (cm)'],
    hue=iris.target, 
    palette='viridis'
)
plt.title('Dispersión con StandardScaler (Media 0, Varianza 1)')
plt.axhline(0, color='black', linestyle='--', alpha=0.3)
plt.axvline(0, color='black', linestyle='--', alpha=0.3)
plt.grid(True, alpha=0.3)
plt.show()
```

### 4. ¿Qué técnica de escalado crees que es más adecuada para este conjunto de datos? ¿Por qué?

```python
Para el dataset Iris, la técnica más adecuada suele ser StandardScaler:

Naturaleza de los datos: Las medidas biológicas (como largos y anchos) suelen seguir una distribución normal o gaussiana, la cual es la base sobre la que trabaja StandardScaler.

Robustez: Si hubiera una flor con una medida errónea (un outlier), MinMaxScaler "aplastaría" todos los datos reales en un rango muy pequeño para dar espacio al error. StandardScaler es más resistente a esto.

Algoritmos: Muchos algoritmos como PCA (Análisis de Componentes Principales) o SVM (Máquinas de Vector de Soporte) asumen explícitamente que los datos están centrados en cero.
```
